Trip Cost Calculator
You are required to design a Python program that helps users calculate the total
cost of a trip, including hotel stay, flight charges, car rental, and additional spending
money.

The program must be modular and use separate functions for each cost
component.

Task 1: Hotel Cost Function

Create a function named hotel_cost that:

• Takes number of nights as input
• Calculates the total hotel cost
• Hotel cost is 140 dollar per night
• Returns the total hotel cost
Task 2: Flight Cost Function
Create a function named plane_ride_cost that:
• Takes a city name (string) as input
• Returns the round-trip flight cost based on the destination
Destination Cost (dollar)
Charlotte 183
Tampa 220
Pittsburgh 222
Los Angeles 475
Task 3: Car Rental Cost Function
Create a function named rental_car_cost that:
• Takes number of rental days as input
• Calculates the base rental cost at 40 dollar per day
• Applies discounts as follows:
o If rental duration is 7 or more days, apply a $50 discount
o Else if rental duration is 3 or more days, apply a 20 dollar discount
o Only one discount can be applied
• Returns the final rental cost
Task 4: Trip Cost Function
Create a function named trip_cost that:
• Takes city and days as input
• Calculates and returns the total trip cost by summing:
o Hotel cost
o Plane ride cost
o Rental car cost
Task 5: Enhanced Trip Cost Calculation
Modify the trip_cost function to:
• Accept a third argument called spending_money
• Add spending_money to the total trip cost
• Return the final amount

```python
#Plane Ride Cost Calculator

def plane_ride_cost(city):
  if city == "Charlotte":
    return 183*2
  elif city == "Tampa":
    return 220*2
  elif city == "Pittsburgh":
    return 222*2
  elif city == "Los Angeles":
    return 475*2
#user output
city =input("Pelase enter the city name").title().strip()
print(f"Selected city is {city} and Your total flight cost will be {plane_ride_cost(city)}$")

```

```python
#Rentel Car Cost Calculator

def rental_car_cost(days):
  cost = 40*days
  if days >= 7:
    cost -= 50
  elif days >= 3:
    cost -= 20
  return cost

rent_days = int(input("Enter the numbers of days you will rent a car"))
print(f"Your total rental car cost will be {rental_car_cost(5)}$")
```
