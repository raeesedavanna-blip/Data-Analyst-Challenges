**Bank Loan Eligibility Criteria **

```python
# Task 1: Input Collection
print("=== Bank Loan Eligibility Portal ===")
age = int(input("Enter your age: "))
income = int(input("Enter your income: "))
credit_score = int(input("Enter your credit score:"))
employemet_status = input("Are you self-employed? (Employed/Unemp ): ").strip().lower()
```

```python
#Loan eligibility conditions
#1. The applicant must be at least 21 years old
if age < 21:
  print(" Rejected : Age criteria not met")
elif income < 20000:
  print(" Rejected: Insufficient income")
elif credit_score < 650:
  print(" Rejected: Credit score criteria not met")
elif employemet_status == "unemployed":
  print(" Rejected: Employement Reuired")
else:
  print(" Loan Approved")

if credit_score >= 750:
  print(" Loan Category: Premium Loan")
else:
  print(" Loan Category: Standard Loan")
```
